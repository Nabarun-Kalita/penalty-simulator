## Verify whether the features look right (in the parquet files)

```python
import pandas as pd
import json

train = pd.read_parquet("C:\\Users\\NABARUN\\Code\\penalty-simulator\\data\\processed\\train_features.parquet")

with open("C:\\Users\\NABARUN\\Code\\penalty-simulator\\data\\processed\\feature_metadata.json") as f:
    meta = json.load(f)

print(f"Train shape: {train.shape}")
print(f"Columns: {len(train.columns)}")
print(f"\nFeature groups:")

for group, cols in meta.items():
    if isinstance(cols, list):
        print(f"  {group}: {len(cols)} columns")

# Verify Messi rows have varying LOO features
messi = train[train['taker_name'].str.contains('Messi', case=False, na=False)]
print(f"\nMessi rows in train: {len(messi)}")
print(f"His LOO conversion range: {messi['taker_conversion_rate_loo'].min():.4f} - {messi['taker_conversion_rate_loo'].max():.4f}")
print(f"His LOO TL preference range: {messi['taker_zone_TL_prob_loo'].min():.4f} - {messi['taker_zone_TL_prob_loo'].max():.4f}")

# Verify a low-reliability player gets near-prior LOO features
print(f"\n--- Low reliability example ---")
lowrel = train[train['taker_total_penalties_loo'] == 0].head(1)

if len(lowrel) > 0:
    print(f"Player: {lowrel.iloc[0]['taker_name']}")
    print(f"LOO conversion rate: {lowrel.iloc[0]['taker_conversion_rate_loo']:.4f}")
    print(f"(Should be ~0.746, the prior)")
```
